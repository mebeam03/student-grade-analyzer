```python

"""
Student Grade Analyzer

Collects student grades, calculates individual averages,
assigns letter grades, and visualizes the distribution
of letter grades using a bar chart.
"""

import matplotlib.pyplot as plt
import seaborn as sns

# Store student names and their grades
grades = {}


def input_scoring():
    num_students = int(input("Enter the number of students: "))

    for i in range(num_students):
        student_name = input(
            f"\nEnter the name of student #{i + 1}: "
        )

        num_grades = int(input(
            f"Enter the number of grades for {student_name}: "
        ))

        student_grades = []

        # Collect grades for each student
        for j in range(num_grades):
            while True:
                grade = float(input(
                    f"Enter grade #{j + 1} for {student_name}: "
                ))

                if 0 <= grade <= 100:
                    student_grades.append(grade)
                    break
                else:
                    print("Grade must be between 0 and 100.")

        grades[student_name] = student_grades

    return grades


def summarize_grades(grades):
    letter_frequencies = {
        'A': 0,
        'B': 0,
        'C': 0,
        'D': 0,
        'F': 0
    }

    print("\n--- Student Grade Summary ---")

    for student, student_grades in grades.items():

        if not student_grades:
            print(f"{student}: No grades entered.")
            continue

        # Calculate average grade
        mean = sum(student_grades) / len(student_grades)

        # Assign letter grade
        if mean >= 90:
            letter = 'A'
        elif mean >= 80:
            letter = 'B'
        elif mean >= 70:
            letter = 'C'
        elif mean >= 60:
            letter = 'D'
        else:
            letter = 'F'

        letter_frequencies[letter] += 1

        print(f"{student}: {mean:.2f}% ({letter})")

    return letter_frequencies


def visualize_grades(letter_frequencies):

    sns.set_style("whitegrid")

    plt.figure(figsize=(9, 5))

    axes = sns.barplot(
        x=list(letter_frequencies.keys()),
        y=list(letter_frequencies.values()),
        color="steelblue"
    )

    axes.set_title("Student Letter Grade Distribution")
    axes.set_xlabel("Letter Grade")
    axes.set_ylabel("Number of Students")

    plt.tight_layout()
    plt.show()


# Run program
student_data = input_scoring()

letter_frequencies = summarize_grades(student_data)

visualize_grades(letter_frequencies)
```
